### What is already known from the companion files
- Photron FASTCAM Mini UX100; image size **1280 × 480**.
- CIHX recording section and MII both report **10,000 fps** and **37,274 images**.
- MII says **uncompressed AVI**; CIHX says **8-bit monochrome export**, with a separate **12-bit effective-depth** field and an image-processing-applied flag.
- CIHX also contains a saved range **0–17,471** and a separate device-state rate **125**. These are not interchangeable with the recording section. The AVI must be checked; do not silently assume a frame mapping.
- The CIHX pixel-size field is **1.0 with an empty unit**. That is not a verified micrometre calibration.

Metadata is evidence, not proof of how the export was produced. This notebook reports disagreements rather than inventing timestamps.

## 3 — Inspect the actual file and preserve the run settings

A unique output folder is created on each run. Keep the original notebook and download intact.

## 4 — View original-resolution sample frames
Six requested AVI indices sample the beginning, middle, and end in nearby pairs (10 frames apart when possible). These are **scouting images, not tracking trajectories**. Each PNG is checked for an exact round-trip of decoded pixels; the contact sheet is a scaled display only. Coordinate labels remain pixels.

Open the PNGs at 100% zoom. Look for: cells visible against background, flow direction, approximate cell diameter, overlaps, blur, illumination changes, and whether a narrower channel crop would retain useful context. Do not infer these from a thumbnail alone.

Seeking is checked against the decoder's reported frame position; this is not independent proof of acquisition timing. The contiguous preview below checks a local sequence.

```python
indices = sample_indices(video['frame_count_header'])
report['samples_complete'] = False
try:
    report['samples'] = export_samples(VIDEO_PATH, RUN_DIR, indices)
    report['samples_complete'] = True
    report.pop('error', None)
    report['status'] = 'samples_complete'
except Exception as e:
    report['status'] = 'sample_error'
    report['error'] = str(e)
    save_report(RUN_DIR, report)
    raise
save_report(RUN_DIR, report)
print('PNG samples saved:', len(report['samples']))
```

## 5 — Optional slow-motion preview
Only a bounded contiguous window is decoded, one frame at a time. The resulting H.264 MP4 is **lossy and for viewing only**. It is not a reduced scientific master and must not be used for speed calculations.

At a confirmed 10,000 fps and one exported frame per capture frame, 200 frames represent roughly **20 milliseconds** of recording, displayed here over **8 seconds**. That may be too short to see a complete cell passage; its purpose is an initial motion check. Adjust the start/window after inspection, rather than reducing frame rate blindly. No sample timestamps are assigned until the export mapping is confirmed.

```python
report['preview'] = {'status': 'disabled'}
if MAKE_PREVIEW:
    n = video['frame_count_header']
    preview_start = max(0, n//2 - PREVIEW_FRAMES//2) if PREVIEW_START is None else PREVIEW_START
    try:
        report['preview'] = make_preview(VIDEO_PATH, RUN_DIR, preview_start, PREVIEW_FRAMES, PREVIEW_PLAYBACK_FPS, n)
        if report['preview']['status'] == 'complete':
            from IPython.display import Video, display
            display(Video(str(RUN_DIR / report['preview']['file']), embed=True, width=960))
    except Exception as e:
        report['preview'] = {'status': 'error', 'error': str(e)}
        print('Preview failed, but metadata and sample PNGs remain available:', e)
save_report(RUN_DIR, report)
print(json.dumps(report['preview'], indent=2))
```

## How we will choose the next algorithm — not implemented yet
- **Clearly separated cells with consistent contrast:** test a simple threshold/connected-components detector and gated centroid assignment.
- **Uneven background:** compare a background-corrected view against the raw frame before adding denoising or thresholding.
- **Touching cells, weak boundaries, or substantial deformation:** the simple detector may fail. Diagnose that before trying more complex segmentation or feature-based linking.
- **Motion blur or frame-mapping uncertainty:** preprocessing cannot recover trustworthy measurements by itself; clarify acquisition/export conditions.

Keep these questions separate: “Can I see/detect each cell?” and “Can I keep the same cell identity across frames?” A preprocessing filter is not a tracker. Final algorithm choice follows the actual images and a manually checked sample, not the file size.
